# SMOTE (Synthetic Minority Over-sampling Technique)

### SMOTE is a technique used to balance an imbalanced dataset by creating new synthetic samples for the minority class

```python
from sklearn.datasets import make_classification
```

```python
## x = independent feature
## y = depended feature
x,y = make_classification(n_samples=1000, n_features=2, n_redundant=0, n_clusters_per_class=1, weights=[0.90], random_state=1)
```

```python
x
```

```python
x,y
```

```python
import pandas as pd
df1= pd.DataFrame(x, columns=['f1','f2'])
df2= pd.DataFrame(y, columns=['target'])
final_df= pd.concat([df1,df2], axis = 1)
```

```python
final_df.head()
```

```python
final_df['target'].value_counts()
```

```python
import matplotlib.pyplot as plt
plt.scatter(final_df['f1'], final_df['f2'], c=final_df['target'])
```

```python
from imblearn.over_sampling import SMOTE
```

```python
# transfer the dataset
oversample=SMOTE()
x,y = oversample.fit_resample(final_df[['f1','f2']],final_df['target'])
```
