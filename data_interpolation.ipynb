# DATA INTERPOLATION

```python
import numpy as np
```

```python
x = np.array([1,2,3,4,5])
y = np.array([2,4,6,8,10])
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(x,y)
```

# interpolate the data using  linear interpolation

```python
x_new = np.linspace(1,5,10) # create new x values
y_interp= np.interp(x_new,x,y) #interpolate y values
print(y_interp)
```

```python
plt.scatter(x_new,y_interp)
```

# cubic interpolation with scipy

```python
import numpy as np
x = np.array([1,2,3,4,5])
y = np.array([1,8,27,64,125])
```

```python
from scipy.interpolate import interp1d
```

### create a cubic interpolation function

```python
f = interp1d(x,y,kind='cubic')
```

```python
# interpolate the data
x_new = np.linspace(1,5,8)
y_interp=f(x_new)
print(y_interp)
```

```python
plt.scatter(x,y)
```

```python
plt.scatter(x_new,y_interp)
```

# polynomial interpolation

```python
import numpy as np

# create some sample data
x = np.array([1,2,3,4,5])
y = np.array([1,4,9,16,25])
```

```python
# interpolate the data using polynomial interpolation
p = np.polyfit(x,y,4) # fit a second degree polynomial to the data
```

```python
x_new = np.linspace(1,5,10) # create new x values
y_interp = np.polyval(p, x_new) # interpolate y values
```

```python
plt.scatter(x,y)
```

```python
plt.scatter(x_new, y_interp)
```
