# Handling Outliers

## 5 number summary
1. Minimum value
2. Q1 - 25 percentile
3. Median
4. Q3 - 75 percentile
5. Maximum

```python
import numpy as np
lst_marks = [45,32,56,75,89,54,32,89,90,87,67,54,45,98,99,67,74,1000,1100]
```

```python
np.percentile(lst_marks,[50])
```

```python
np.percentile(lst_marks,[25])
```

```python
np.percentile(lst_marks,[75])
```

```python
##[lower fence -----> higher fence]
Q1=np.percentile(lst_marks,[25])
print(Q1)
```

```python
minimum,Q1,Q2,Q3,maximum=np.quantile(lst_marks,[0,0.25,0.50,0.75,1.0])
```

```python
IQR =Q3-Q1
print(IQR)
```

```python
lower_fence = Q1-1.5*(IQR)
higher_fence = Q3+1.5*(IQR)
```
