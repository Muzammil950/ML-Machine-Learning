# Covariance and Correlation

```python
import seaborn as sns
```

```python
df = sns.load_dataset('healthexp')
df.head()
```

```python
df_cov = df.select_dtypes(include=['number'])
df_cov.cov()
```

```python
# Pearson correlation coefficient
df_corr_prsn = df.select_dtypes(include=['number'])
df_corr_prsn.corr(method='pearson')
```

```python
## spearman rank correlation
df_corr_sprn = df.select_dtypes(include=['number'])
df_corr_sprn.corr(method='spearman')
```
