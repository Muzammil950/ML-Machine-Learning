# Label And Ordinal Encoding

```python
import pandas as pd
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
# create an sample dataframe for categorical variable
df = pd.DataFrame({
    'colour' : ['red','blue','green','green','red','blue']
    })
```

```python
df
```

```python
# create an instance of labelencoder
encoder = LabelEncoder()
```

```python
encoder.fit_transform(df['colour'])
```

```python
# Ordinal Encoding
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

# create a sample dataframe for categorical variables
df = pd.DataFrame({
    'size' : ['small','medium','large','medium','small','large']
       })
```

```python
## create an instance of the Ordinal Encoder class and fit_transform
encoder= OrdinalEncoder(categories= [['small','medium','large']])
                                      
```

```python
encoder.fit_transform(df[['size']])
```
